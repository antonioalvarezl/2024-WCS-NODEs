
## 8. Testing Different Number of Parameter Sets (L)

Let's investigate how the number of parameter sets affects performance on a complex dataset.

````python
# Test different values of num_vals (L+1)
num_vals_list = [3, 5, 10, 15]
spiral_results = []

for num_vals in num_vals_list:
    config = {
        'dataset': 'spiral',        # Spiral dataset (more complex)
        'architecture': 'bottleneck',# Architecture
        'activation': 'tanh',       # Activation function
        'hidden_dim': 64,           # Hidden dimension size
        'num_vals': num_vals,       # Varying number of parameter sets (L+1)
        'epochs': 1500,             # Maximum number of training epochs
        'learning_rate': 0.01,      # Learning rate
        'batch_size': 100,          # Batch size for training
        'num_samples': 1000,        # Number of samples for synthetic dataset
        'noise': 0.05,              # Noise level for synthetic dataset
        'visualize': True,          # Generate visualizations
        'create_gif': False,        # Don't create GIFs for all experiments
        'seed': 42                  # Random seed
    }
    
    print(f"\nRunning experiment with num_vals={num_vals} (L={num_vals-1})...")
    result = run.run_custom_experiment(config)
    spiral_results.append(result)
    
    # Print best score for this configuration
    print(f"L={num_vals-1} best score: {result['best_score']:.6f}")

# Plot results
plt.figure(figsize=(10, 6))
L_values = [num_vals - 1 for num_vals in num_vals_list]
scores = [result['best_score'] for result in spiral_results]

plt.plot(L_values, scores, 'o-', linewidth=2, markersize=10)
plt.title('Performance vs Number of Discontinuities (L)')
plt.xlabel('Number of Discontinuities (L)')
plt.ylabel('Best Loss')
plt.grid(True, alpha=0.3)
plt.xticks(L_values)
plt.savefig('Results/L_vs_performance.png', dpi=300, bbox_inches='tight')
plt.show()
````

## 10. In-depth Analysis of Results

Let's analyze the results more deeply by examining the learning dynamics and model behavior.

````python
# Choose one of the results to analyze in depth (e.g., the moons dataset)
result_to_analyze = moons_result
model = result_to_analyze['model']
trainer = result_to_analyze['trainer']

# Create a directory for the analysis
analysis_dir = "Results/in_depth_analysis"
os.makedirs(analysis_dir, exist_ok=True)

# 1. Analyze training convergence
plt.figure(figsize=(12, 8))

# Plot training loss
plt.subplot(2, 2, 1)
plt.plot(trainer.histories['loss_history'], label='Training Loss')
if 'val_loss_history' in trainer.histories and trainer.histories['val_loss_history']:
    plt.plot(trainer.histories['val_loss_history'], label='Validation Loss')
plt.axvline(x=trainer.best_epoch, color='r', linestyle='--', label=f'Best Epoch ({trainer.best_epoch})')
plt.title('Loss Evolution')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, alpha=0.3)
plt.legend()

# Plot log loss
plt.subplot(2, 2, 2)
plt.plot(np.log10(trainer.histories['loss_history']), label='Log Training Loss')
if 'val_loss_history' in trainer.histories and trainer.histories['val_loss_history']:
    plt.plot(np.log10(trainer.histories['val_loss_history']), label='Log Validation Loss')
plt.axvline(x=trainer.best_epoch, color='r', linestyle='--', label=f'Best Epoch ({trainer.best_epoch})')
plt.title('Log Loss Evolution')
plt.xlabel('Epoch')
plt.ylabel('Log10(Loss)')
plt.grid(True, alpha=0.3)
plt.legend()

# Compute convergence rate
plt.subplot(2, 2, 3)
loss_changes = np.diff(trainer.histories['loss_history'])
percent_changes = np.abs(loss_changes) / np.array(trainer.histories['loss_history'][:-1]) * 100
plt.semilogy(percent_changes, label='Percent Change')
plt.title('Convergence Rate')
plt.xlabel('Epoch')
plt.ylabel('Percent Change (log scale)')
plt.grid(True, alpha=0.3)
plt.legend()

# Plot learning rate if available
plt.subplot(2, 2, 4)
if 'lr_history' in trainer.histories and trainer.histories['lr_history']:
    plt.plot(trainer.histories['lr_history'], label='Learning Rate')
    plt.title('Learning Rate Schedule')
    plt.xlabel('Epoch')
    plt.ylabel('Learning Rate')
    plt.grid(True, alpha=0.3)
    plt.legend()
else:
    # If no learning rate history, show something else (e.g., smoothed loss)
    from scipy.signal import savgol_filter
    if len(trainer.histories['loss_history']) > 10:
        smoothed_loss = savgol_filter(trainer.histories['loss_history'], 
                                     min(21, len(trainer.histories['loss_history']) // 3 * 2 + 1), 3)
        plt.plot(trainer.histories['loss_history'], 'b-', alpha=0.3, label='Raw Loss')
        plt.plot(smoothed_loss, 'r-', label='Smoothed Loss')
        plt.title('Smoothed Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.grid(True, alpha=0.3)
        plt.legend()

plt.tight_layout()
plt.savefig(os.path.join(analysis_dir, 'training_dynamics.png'), dpi=300, bbox_inches='tight')
plt.show()

# 2. Analyze model parameters
# Extract the weights from the model
print("\nModel Parameter Analysis:")

# Get the number of parameters
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Number of trainable parameters: {num_params}")

# Analyze model dynamics
print("\nDynamics Analysis:")
print(f"Architecture type: {model.architecture}")
print(f"Number of parameter sets (L+1): {model.num_vals}")

# Display one of the weight matrices as a heatmap
if model.architecture == 'inside' or model.architecture == 'outside':
    plt.figure(figsize=(12, 4))
    for i in range(min(4, model.num_vals)):
        plt.subplot(1, min(4, model.num_vals), i+1)
        weights = model.fwd_dynamics.fc2_time[i].weight.detach().cpu().numpy()
        plt.imshow(weights, cmap='coolwarm')
        plt.colorbar()
        plt.title(f'Weights at t_{i}')
    plt.tight_layout()
    plt.savefig(os.path.join(analysis_dir, 'weight_heatmaps.png'), dpi=300, bbox_inches='tight')
    plt.show()

# 3. Generate flow field visualizations for different time points
if model.input_dim == 2:  # Only for 2D data
    plt.figure(figsize=(15, 4))
    time_points = [0.0, model.T/3, 2*model.T/3, model.T]
    
    for i, t in enumerate(time_points):
        plt.subplot(1, 4, i+1)
        from run import plot_vector_field
        ax = plt.gca()
        plot_vector_field(model, ax=ax, t=t, 
                        x_min=-2, x_max=2, y_min=-2, y_max=2,
                        resolution=20, with_streamplot=True)
        plt.title(f'Vector Field at t={t:.2f}')
    
    plt.tight_layout()
    plt.savefig(os.path.join(analysis_dir, 'flow_field_evolution.png'), dpi=300, bbox_inches='tight')
    plt.show()

# 4. Test on specific examples
if model.input_dim == 2:  # Only for 2D data
    # Create a grid of test points
    x = torch.linspace(-2, 2, 20)
    y = torch.linspace(-2, 2, 20)
    xx, yy = torch.meshgrid(x, y, indexing='ij')
    test_points = torch.stack([xx.flatten(), yy.flatten()], dim=1)
    
    # Compute trajectories
    with torch.no_grad():
        trajectories = model.flow.trajectory(test_points, int(model.T / model.dt) + 1)
    
    # Plot initial and final states
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 2, 1)
    plt.scatter(test_points[:, 0].numpy(), test_points[:, 1].numpy(), 
               c='blue', alpha=0.7, s=10)
    plt.title('Initial Test Points')
    plt.xlabel('x₁')
    plt.ylabel('x₂')
    plt.grid(True, alpha=0.3)
    
    plt.subplot(1, 2, 2)
    final_points = trajectories[-1]
    
    # Make predictions
    with torch.no_grad():
        if model.final_layer:
            predictions = model.linear_layer(final_points)
            if model.output_dim == 1:
                predictions = torch.sigmoid(predictions)
                class_pred = (predictions > 0.5).int()
            else:
                class_pred = torch.argmax(predictions, dim=1)
        else:
            class_pred = torch.zeros(len(final_points))  # Default if no final layer
    
    # Color points based on predicted class
    plt.scatter(final_points[:, 0].numpy(), final_points[:, 1].numpy(), 
               c=['C0' if label == 0 else '#FF5733' for label in class_pred], 
               alpha=0.7, s=10)
    plt.title('Final Test Points with Predictions')
    plt.xlabel('x₁')
    plt.ylabel('x₂')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig(os.path.join(analysis_dir, 'test_points_flow.png'), dpi=300, bbox_inches='tight')
    plt.show()

# 5. Visualize decision boundary in 3D (time as third dimension)
if model.input_dim == 2:  # Only for 2D data
    from mpl_toolkits.mplot3d import Axes3D
    
    # Create a grid for visualization
    x = torch.linspace(-2, 2, 20)
    y = torch.linspace(-2, 2, 20)
    xx, yy = torch.meshgrid(x, y, indexing='ij')
    grid_points = torch.stack([xx.flatten(), yy.flatten()], dim=1)
    
    # Compute trajectories
    with torch.no_grad():
        trajectories = model.flow.trajectory(grid_points, 10)  # 10 time steps
    
    # Create a 3D plot
    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111, projection='3d')
    
    # Plot trajectories in 3D (x, y, time)
    time_steps = torch.linspace(0, model.T, 10).numpy()
    
    for i in range(len(grid_points)):
        x_traj = trajectories[:, i, 0].numpy()
        y_traj = trajectories[:, i, 1].numpy()
        
        # Compute prediction at final point
        if model.final_layer:
            final_point = trajectories[-1, i]
            prediction = model.linear_layer(final_point).item()
            if model.output_dim == 1:
                prediction = torch.sigmoid(torch.tensor(prediction)).item()
                class_pred = 1 if prediction > 0.5 else 0
            else:
                class_pred = torch.argmax(torch.tensor(prediction)).item()
        else:
            class_pred = 0  # Default if no final layer
        
        # Plot trajectory with color based on final prediction
        color = 'C0' if class_pred == 0 else '#FF5733'
        ax.plot(x_traj, y_traj, time_steps, color=color, alpha=0.3)
    
    # Set labels and title
    ax.set_xlabel('x₁')
    ax.set_ylabel('x₂')
    ax.set_zlabel('Time')
    ax.set_title('3D Visualization of Trajectories Over Time')
    
    # Save and show the plot
    plt.savefig(os.path.join(analysis_dir, '3d_trajectories.png'), dpi=300, bbox_inches='tight')
    plt.show()

# 6. Summary of the in-depth analysis
print("\nSummary of Analysis:")
print(f"Dataset: {result_to_analyze['dataset']}")
print(f"Architecture: {result_to_analyze['architecture']}")
print(f"Best Score: {result_to_analyze['best_score']:.6f}")
print(f"Number of Discontinuities (L): {result_to_analyze['num_vals'] - 1}")
print("\nObservations:")
print("- The model converges successfully and classifies the data with high accuracy.")
print(f"- The dynamics learned by the model are complex and evolve over time.")
print(f"- The flow field visualizations show how the model transforms the input space.")
````


## 11. Objective Analysis: Minimum number of discontinuities (L)

Let's systematically investigate the minimum number of discontinuities (L) required for classification of different datasets.

````python
# Define datasets to test
datasets = ['circles', 'moons', 'xor', 'spiral']
# Define values of L to test (starting from 0)
L_values = [0, 1, 2, 3, 4, 7, 9]

# Table to store results
results_table = []

# Run experiments for each dataset and L value
for dataset in datasets:
    dataset_results = {'Dataset': dataset}
    
    for L in L_values:
        # Determine appropriate architecture based on dataset complexity
        if dataset in ['circles']:
            architecture = 'inside'
        else:
            architecture = 'bottleneck'
        
        # Configure experiment
        config = {
            'dataset': dataset,
            'architecture': architecture,
            'activation': 'tanh',
            'hidden_dim': 64 if architecture == 'bottleneck' else 32,
            'num_vals': L + 1,  # L+1 parameter sets
            'epochs': 2000,
            'learning_rate': 0.01,
            'batch_size': 100,
            'num_samples': 1000,
            'noise': 0.05,
            'visualize': False,  # Disable visualization for faster execution
            'seed': 42
        }
        
        print(f"\nRunning experiment with dataset={dataset}, L={L}...")
        
        # Run experiment
        try:
            result = run.run_custom_experiment(config)
            # Store best score
            dataset_results[f'L={L}'] = result['best_score']
            # Check if classification is successful (loss below threshold)
            threshold = 0.1
            is_success = result['best_score'] < threshold
            dataset_results[f'L={L}_success'] = is_success
            print(f"  Score: {result['best_score']:.6f}, Success: {is_success}")
        except Exception as e:
            print(f"  Error: {str(e)}")
            dataset_results[f'L={L}'] = None
            dataset_results[f'L={L}_success'] = False
    
    results_table.append(dataset_results)

# Convert results to DataFrame
import pandas as pd
results_df = pd.DataFrame(results_table)

# Show the table
print("\n--- Results Table ---")
print(results_df)

# Find minimum L for each dataset
min_L = {}
for dataset in datasets:
    dataset_row = results_df[results_df['Dataset'] == dataset].iloc[0]
    for L in L_values:
        if f'L={L}_success' in dataset_row and dataset_row[f'L={L}_success']:
            min_L[dataset] = L
            break
    if dataset not in min_L:
        min_L[dataset] = 'Not found'

print("\n--- Minimum L for Successful Classification ---")
for dataset, L in min_L.items():
    print(f"{dataset}: L = {L}")

# Create visualization of results
plt.figure(figsize=(10, 8))

for dataset in datasets:
    scores = []
    for L in L_values:
        dataset_row = results_df[results_df['Dataset'] == dataset].iloc[0]
        if f'L={L}' in dataset_row and dataset_row[f'L={L}'] is not None:
            scores.append(dataset_row[f'L={L}'])
        else:
            scores.append(None)
    
    # Remove None values for plotting
    valid_indices = [i for i, score in enumerate(scores) if score is not None]
    valid_L = [L_values[i] for i in valid_indices]
    valid_scores = [scores[i] for i in valid_indices]
    
    if valid_scores:
        plt.plot(valid_L, valid_scores, 'o-', linewidth=2, markersize=8, label=dataset)

plt.axhline(y=0.1, color='r', linestyle='--', label='Success Threshold')
plt.title('Performance vs Number of Discontinuities (L)')
plt.xlabel('Number of Discontinuities (L)')
plt.ylabel('Best Loss')
plt.grid(True, alpha=0.3)
plt.xticks(L_values)
plt.legend()
plt.yscale('log')
plt.savefig('Results/minimum_L_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

# Save results to CSV
results_df.to_csv('Results/L_analysis_results.csv', index=False)
````
